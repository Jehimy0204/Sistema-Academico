```python
class Alumno:
    def __init__(self, cedula, nombre, edad, nota):
        self.__cedula = cedula
        self.__nombre = nombre
        self.__edad = edad
        self.__nota = nota

    def get_cedula(self):
        return self.__cedula

    def get_nombre(self):
        return self.__nombre

    def get_edad(self):
        return self.__edad

    def get_nota(self):
        return self.__nota

    def set_nombre(self, nombre):
        self.__nombre = nombre

    def set_edad(self, edad):
        self.__edad = edad

    def set_nota(self, nota):
        self.__nota = nota

    def __str__(self):
        return f"Alumno: {self.__nombre} | Cédula: {self.__cedula} | Edad: {self.__edad} | Nota: {self.__nota}"

    def puede_votar(self):
        return self.__edad >= 18

    def obtener_calificacion(self):
        return "Aprobado" if self.__nota >= 3.0 else "Reprobado"
```

```python
class Materia:
    def __init__(self, nombre, creditos, docente):
        self.nombre = nombre
        self.creditos = creditos
        self.docente = docente
        self.estudiantes = []

    def asignar_alumno(self, alumno):
        if isinstance(alumno, Alumno):
            self.estudiantes.append(alumno)

    def listar_estudiantes(self):
        print(f"\nEstudiantes en la materia: {self.nombre}")
        for est in self.estudiantes:
            print(est)
            print(f"Puede votar: {'Sí' if est.puede_votar() else 'No'}")
            print(f"Estado: {est.obtener_calificacion()}")
            print("-" * 40)
```

```python
class SistemaAcademico:
    def __init__(self):
        self.materias = []

    def crear_materia(self, nombre, creditos, docente):
        materia = Materia(nombre, creditos, docente)
        self.materias.append(materia)
        return materia

    def ver_materias(self):
        print("\nMaterias registradas:")
        for mat in self.materias:
            print(f"{mat.nombre} | Docente: {mat.docente} | Créditos: {mat.creditos}")

    def resumen_materias(self):
        for materia in self.materias:
            materia.listar_estudiantes()
```

```python
# Crear sistema académico
sistema = SistemaAcademico()

# Crear materias
mat1 = sistema.crear_materia("Matemáticas", 4, "Profesor García")
mat2 = sistema.crear_materia("Física", 3, "Profesora López")

# Crear alumnos
al1 = Alumno("101", "Ana", 19, 4.5)
al2 = Alumno("102", "Luis", 17, 2.8)
al3 = Alumno("103", "Carlos", 20, 3.2)

# Asignar alumnos a materias
mat1.asignar_alumno(al1)
mat1.asignar_alumno(al2)
mat2.asignar_alumno(al3)
mat2.asignar_alumno(al1)

# Mostrar información
sistema.ver_materias()
sistema.resumen_materias()
```
